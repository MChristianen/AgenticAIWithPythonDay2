```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
```

```python
from getpass import getpass
os.environ['OPENAI_API_KEY'] = getpass('Voer je OpenAI API key in: ')
```

```python
llm = ChatOpenAI(
    api_key=os.getenv('OPENAI_API_KEY'),
    model=os.getenv('OPENAI_MODEL', 'gpt-4o-mini'),
)
```

```python
prompt = "Why is gas so expensive?"
style  = "poetic"   # try: formal / punchy / simple / gen-z
```

```python
# ── Step 1: elaborate ────────────────────────────────────────────────
step1      = llm.invoke(f"""
You are a creative writer. Write a detailed, elaborate response to:
{prompt}
Be thorough and expressive.
""")
elaborated = step1.content
```

```python
# ── Step 2: condense + stylize ───────────────────────────────────────
step2     = llm.invoke(f"""
Rewrite the text below in a {style} style.
Maximum 100 words. Keep the core message intact.

Text: {elaborated}
""")
condensed = step2.content
```

```python
# ── Step 3: verify ───────────────────────────────────────────────────
step3   = llm.invoke(f"""
Does this condensed text still answer the original prompt?
Original prompt : {prompt}
Condensed text  : {condensed}
Reply PASS or FAIL, then one sentence explaining why.
""")
verdict = step3.content
```

```python
# ── Output ───────────────────────────────────────────────────────────
print("=== Condensed ===")
print(condensed)
print("\n=== Verdict ===")
print(verdict)
```
